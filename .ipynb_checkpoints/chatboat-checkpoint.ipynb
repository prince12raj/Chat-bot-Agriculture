```python
import pandas as pd

file_path = r"C:\Users\prince raj\ChatBoat For Agriculture\datasets\csv.csv"

df = pd.read_csv(file_path)

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nDataset Info:")
df.info()
```

```python
qa_data = []

for index, row in df.iterrows():

    commodity = row['commodity']
    state = row['state']
    district = row['district']
    market = row['market']
    variety = row['variety']

    min_price = row['min_price']
    max_price = row['max_price']
    modal_price = row['modal_price']

    arrival_date = row['arrival_date']

    # Question
    question = (
        f"What is the price of {commodity} "
        f"in {market} market of {district}, {state}?"
    )

    # Answer
    answer = (
        f"Commodity: {commodity}\n"
        f"Variety: {variety}\n"
        f"State: {state}\n"
        f"District: {district}\n"
        f"Market: {market}\n"
        f"Arrival Date: {arrival_date}\n"
        f"Minimum Price: ₹{min_price}\n"
        f"Maximum Price: ₹{max_price}\n"
        f"Modal Price: ₹{modal_price}"
    )

    qa_data.append([question, answer])

print("Question Answer Data Created")
```

```python
qa_df = pd.DataFrame(
    qa_data,
    columns=["question", "answer"]
)

qa_df.head()
```

```python
save_path = r"C:\Users\prince raj\ChatBoat For Agriculture\datasets\market_chatbot_data.csv"

qa_df.to_csv(save_path, index=False)

print("Chatbot dataset saved")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    'all-MiniLM-L6-v2'
)

print("AI Model Loaded")
```

```python
questions = qa_df['question'].tolist()

question_embeddings = model.encode(
    questions,
    show_progress_bar=True
)

print("Embeddings Created")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

import numpy as np
```

```python
def agriculture_chatbot(user_query):

    query_embedding = model.encode([user_query])

    similarities = cosine_similarity(
        query_embedding,
        question_embeddings
    )

    best_match_index = np.argmax(similarities)

    response = qa_df.iloc[
        best_match_index
    ]['answer']

    return response
```

```python
query = "What is tomato price in market?"

response = agriculture_chatbot(query)

print(response)
```

```python
print("Agriculture Market Chatbot Started")
print("Type exit to stop")

while True:

    user_input = input("delhi ")

    if user_input.lower() == "exit":

        print("Chatbot Closed")

        break

    response = agriculture_chatbot(user_input)

    print("\nBot:\n")

    print(response)

    print("\n")
```

```python
csv_files = glob.glob(
    r"C:\Users\prince raj\ChatBoat For Agriculture\datasets\**\*.csv",
    recursive=True
)

print("Total CSV Files:", len(csv_files))

for file in csv_files:
    
    print(file)
```

```python
dataframes = []

for file in csv_files:
    
    try:
        
        df = pd.read_csv(file)
        
        df['source_file'] = os.path.basename(file)
        
        dataframes.append(df)
        
        print("Loaded:", file)
        
    except Exception as e:
        
        print("Error:", file)
        
        print(e)
```

```python
qa_data = []

for df in dataframes:
    
    columns = list(df.columns)
    
    for index, row in df.head(1000).iterrows():
        
        question = ""
        
        answer = ""
        
        # Dynamic Question Creation
        if 'label' in columns:
            
            question = (
                f"Which crop is suitable "
                f"for soil and weather conditions?"
            )
            
        elif 'Temparature' in columns or 'Temperature' in columns:
            
            question = (
                f"What is the weather information?"
            )
            
        elif 'Fertilizer Name' in columns:
            
            question = (
                f"Which fertilizer should be used?"
            )
            
        else:
            
            question = (
                f"Give agriculture information"
            )
        
        # Dynamic Answer Creation
        for col in columns:
            
            answer += f"{col}: {row[col]}\n"
        
        qa_data.append([question, answer])

print("Unified QA Data Created")

print("Total QA Pairs:", len(qa_data))
```

```python
qa_df = pd.DataFrame(
    qa_data,
    columns=["question", "answer"]
)

qa_df.head()
```

```python
save_path = r"C:\Users\prince raj\ChatBoat For Agriculture\datasets\final_data\agriculture_chatbot_dataset.csv"

qa_df.to_csv(save_path, index=False)

print("Final chatbot dataset saved")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    'all-MiniLM-L6-v2'
)

print("AI Model Loaded Successfully")
```

```python
questions = qa_df['question'].tolist()

question_embeddings = model.encode(
    questions,
    show_progress_bar=True
)

print("Embeddings Created")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

import numpy as np
```

```python
def agriculture_chatbot(user_query):
    
    query_embedding = model.encode([user_query])
    
    similarities = cosine_similarity(
        query_embedding,
        question_embeddings
    )
    
    best_match_index = np.argmax(similarities)
    
    response = qa_df.iloc[
        best_match_index
    ]['answer']
    
    return response
```

```python
query = "Best fertilizer for crops"

response = agriculture_chatbot(query)

print(response)
```

```python
print("Agriculture AI Chatbot Started")
print("Type exit to stop")

while True:
    
    user_input = input("You: ")
    
    if user_input.lower() == "exit":
        
        print("Chatbot Closed")
        
        break
    
    response = agriculture_chatbot(user_input)
    
    print("\nBot:\n")
    
    print(response)
    
    print("\n")
```
