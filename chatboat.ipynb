```python
import pandas as pd

file_path = r"C:\Users\prince raj\ChatBoat For Agriculture\datasets\csv.csv"

df = pd.read_csv(file_path)

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nDataset Info:")
df.info()
```

```python
qa_data = []

for index, row in df.iterrows():

    commodity = row['commodity']
    state = row['state']
    district = row['district']
    market = row['market']
    variety = row['variety']

    min_price = row['min_price']
    max_price = row['max_price']
    modal_price = row['modal_price']

    arrival_date = row['arrival_date']

    # Question
    question = (
        f"What is the price of {commodity} "
        f"in {market} market of {district}, {state}?"
    )

    # Answer
    answer = (
        f"Commodity: {commodity}\n"
        f"Variety: {variety}\n"
        f"State: {state}\n"
        f"District: {district}\n"
        f"Market: {market}\n"
        f"Arrival Date: {arrival_date}\n"
        f"Minimum Price: ₹{min_price}\n"
        f"Maximum Price: ₹{max_price}\n"
        f"Modal Price: ₹{modal_price}"
    )

    qa_data.append([question, answer])

print("Question Answer Data Created")
```

```python
qa_df = pd.DataFrame(
    qa_data,
    columns=["question", "answer"]
)

qa_df.head()
```

```python
save_path = r"C:\Users\prince raj\ChatBoat For Agriculture\datasets\market_chatbot_data.csv"

qa_df.to_csv(save_path, index=False)

print("Chatbot dataset saved")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    'all-MiniLM-L6-v2'
)

print("AI Model Loaded")
```

```python
questions = qa_df['question'].tolist()

question_embeddings = model.encode(
    questions,
    show_progress_bar=True
)

print("Embeddings Created")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

import numpy as np
```

```python
def agriculture_chatbot(user_query):

    query_embedding = model.encode([user_query])

    similarities = cosine_similarity(
        query_embedding,
        question_embeddings
    )

    best_match_index = np.argmax(similarities)

    response = qa_df.iloc[
        best_match_index
    ]['answer']

    return response
```

```python
query = "What is tomato price in market?"

response = agriculture_chatbot(query)

print(response)
```

```python
dataframes = []

for file in csv_files:
    
    try:
        
        df = pd.read_csv(file)
        
        df['source_file'] = os.path.basename(file)
        
        dataframes.append(df)
        
        print("Loaded:", file)
        
    except Exception as e:
        
        print("Error loading:", file)
        
        print(e)
```

```python
qa_data = []

for df in dataframes:
    
    columns = list(df.columns)

    for index, row in df.head(3000).iterrows():

        answer = ""

        # =========================================
        # WEATHER DATASET
        # =========================================
        
        if (
            'Temperature' in columns or
            'Temparature' in columns or
            'Humidity' in columns
        ):

            question = (
                f"What is the weather information?"
            )

        # =========================================
        # CROP RECOMMENDATION
        # =========================================

        elif (
            'label' in columns or
            'N' in columns and
            'P' in columns and
            'K' in columns
        ):

            question = (
                f"Which crop is suitable "
                f"for this soil condition?"
            )

        # =========================================
        # FERTILIZER DATASET
        # =========================================

        elif (
            'Fertilizer Name' in columns or
            'Fertilizer' in columns
        ):

            question = (
                f"Which fertilizer should I use?"
            )

        # =========================================
        # MARKET PRICE DATASET
        # =========================================

        elif (
            'commodity' in columns or
            'modal_price' in columns
        ):

            commodity = row.get(
                'commodity',
                'crop'
            )

            question = (
                f"What is the market price "
                f"of {commodity}?"
            )

        # =========================================
        # SMART FARMING DATASET
        # =========================================

        else:

            question = (
                f"Give smart farming information"
            )

        # =========================================
        # CREATE DETAILED ANSWER
        # =========================================

        for col in columns:

            answer += (
                f"{col}: {row[col]}\n"
            )

        qa_data.append([
            question,
            answer
        ])

print("FINAL QA DATA CREATED")

print("Total QA Pairs:", len(qa_data))
```

```python
csv_files = glob.glob(
    r"C:\Users\prince raj\ChatBoat For Agriculture\datasets\**\*.csv",
    recursive=True
)

print(csv_files)
```

```python
dataframes = []

for file in csv_files:

    try:

        df = pd.read_csv(file)

        df['source_file'] = os.path.basename(file)

        dataframes.append(df)

        print("Loaded:", file)

    except Exception as e:

        print(e)
```

```python
qa_data = []

for df in dataframes:

    columns = list(df.columns)

    for index, row in df.head(2000).iterrows():

        answer = ""

        # =========================================
        # MARKET PRICE DATASET
        # =========================================

        if (
            'commodity' in columns and
            'modal_price' in columns
        ):

            commodity = row.get('commodity', '')

            state = row.get('state', '')

            market = row.get('market', '')

            question = (
                f"What is the price of "
                f"{commodity} in {market} market?"
            )

        # =========================================
        # CROP RECOMMENDATION DATASET
        # =========================================

        elif (
            'N' in columns and
            'P' in columns and
            'K' in columns and
            'label' in columns
        ):

            question = (
                f"Which crop grows best "
                f"with NPK values "
                f"{row['N']}, {row['P']}, {row['K']}?"
            )

        # =========================================
        # FERTILIZER DATASET
        # =========================================

        elif (
            'Fertilizer Name' in columns
        ):

            crop = row.get('Crop Type', '')

            soil = row.get('Soil Type', '')

            question = (
                f"Best fertilizer for "
                f"{crop} crop in {soil} soil?"
            )

        # =========================================
        # WEATHER DATASET
        # =========================================

        elif (
            'Temperature' in columns or
            'Temparature' in columns
        ):

            temp = row.get(
                'Temperature',
                row.get('Temparature', '')
            )

            humidity = row.get(
                'Humidity',
                ''
            )

            question = (
                f"What weather has "
                f"temperature {temp} "
                f"and humidity {humidity}?"
            )

        # =========================================
        # SMART FARMING
        # =========================================

        else:

            question = (
                "Give smart farming advice"
            )

        # =========================================
        # CREATE ANSWER
        # =========================================

        for col in columns:

            answer += (
                f"{col}: {row[col]}\n"
            )

        qa_data.append([
            question,
            answer
        ])

print("SMART QA DATASET CREATED")

print("Total QA:", len(qa_data))
```
